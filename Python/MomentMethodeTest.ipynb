```python
import numpy as np
from sympy import symbols, Matrix
```

```python
A, B, C, D, E, F = symbols('A B C D E F')
a, b, c, d, e, f = symbols('a b c d e f')

X = Matrix([
    [A,B],
    [C,D],
    [E,F]
])
Y = Matrix([
    [a,c,e],
    [b,d,f]
])
S=X*Y
S
```

```python
np.trace(S**3)
```

```python
TestGraph = (A*a+B*b)*(A*e+B*f)*(C*a+D*d)*(C*c+D*d)*(E*c+F*d)*(E*e+F*f)
```

```python
if TestGraph == np.trace(S**3):
    print('ist leider gleich')
else: 
    print('juhu ich lag richtig (ist nicht gleich)')
```
